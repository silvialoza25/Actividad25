 Crear un análisis simple de las ventas de un pequeño negocio, incluyendo la
creación de un DataFrame, el uso de operaciones de NumPy y la visualización de los resultados
con Matplotlib.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Datos de ventas mensuales (en miles)
data = {"Mes": ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio", "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"], "Ventas": [25, 30, 28, 35, 40, 50, 55, 60, 70, 80, 75, 90]}

# Crear un DataFrame
df_ventas = pd.DataFrame(data)

print("DataFrame de Ventas:")
display(df_ventas)

# Operaciones de NumPy
ventas_array = df_ventas['Ventas'].to_numpy()
media_ventas = np.mean(ventas_array)
max_ventas = np.max(ventas_array)
min_ventas = np.min(ventas_array)

print(f"\nEstadísticas de Ventas (en miles):\nMedia: {media_ventas:.2f}\nMáximo: {max_ventas}\nMínimo: {min_ventas}")
```

```python
# Visualización de los resultados con Matplotlib
plt.figure(figsize=(12, 6))
plt.bar(df_ventas['Mes'], df_ventas['Ventas'], color='skyblue')
plt.xlabel('Mes')
plt.ylabel('Ventas (en miles)')
plt.title('Ventas Mensuales del Negocio')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
