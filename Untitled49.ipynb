```python
import random  # Paso 1: Importar el módulo random


# Paso 2: Función que retorna un número aleatorio entre 1 y 100
def generar_numero():
    return random.randint(1, 100)


def jugar():
    numero_secreto = generar_numero()
    intentos = 0  # Paso 6: Contador de intentos

    print("=" * 40)
    print("   🎯 JUEGO: ADIVINA EL NÚMERO")
    print("=" * 40)
    print("He pensado un número entre 1 y 100.")
    print("¿Puedes adivinarlo?\n")

    # Paso 3: Bucle while para el juego principal
    while True:

        # Paso 4: try-except para manejar entradas no numéricas
        try:
            intento = int(input("Ingresa tu número: "))
        except ValueError:
            print("⚠️  Entrada inválida. Por favor ingresa un número entero.\n")
            continue

        intentos += 1  # Incrementar contador tras cada intento válido

        # Paso 5: if-elif-else para dar pistas al jugador
        if intento < numero_secreto:
            print(f"📈 El número secreto es MAYOR que {intento}. ¡Sigue intentando!\n")
        elif intento > numero_secreto:
            print(f"📉 El número secreto es MENOR que {intento}. ¡Sigue intentando!\n")
        else:
            print(f"\n✅ ¡CORRECTO! El número era {numero_secreto}.")
            print(f"🏆 Lo lograste en {intentos} intento(s).")
            print("=" * 40)
            break  # Paso 7: break para salir del bucle al adivinar


# Punto de entrada del programa
if __name__ == "__main__":
    jugar()
```
